```python
# PySpark-Optimierte Hop-Variabilität-Analyse: Messpunkt-Zielservice-Paare
# ==================================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

# PySpark Imports
from pyspark.sql import SparkSession, functions as F, types as T
from pyspark.sql.functions import (
    col, when, isnan, isnull, size, expr, explode, collect_list, 
    avg, count, stddev, min as spark_min, max as spark_max,
    window, date_format, hour, dayofweek, date_trunc,
    udf, struct, array, lit, coalesce
)
from pyspark.sql.types import (
    StructType, StructField, StringType, IntegerType, DoubleType, 
    TimestampType, ArrayType, BooleanType, FloatType
)

# Statistische Analysen
from scipy import stats
from collections import defaultdict, Counter
import json
from math import radians, cos, sin, asin, sqrt

# Konfiguration
plt.style.use('default')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (15, 10)

print("=" * 120)
print("🚀 PYSPARK-OPTIMIERTE HOP-VARIABILITÄT-ANALYSE")
print("Hochperformante Routing-Stabilität & Pfad-Variabilität für 140 Messpunkt-Service-Kombinationen")
print("=" * 120)

# ================================================================
# 1. SPARK SESSION UND KONFIGURATION
# ================================================================

def create_optimized_spark_session():
    """Erstellt optimierte Spark Session für Hop-Variabilität-Analyse"""
    
    spark = SparkSession.builder \
        .appName("MTR_Hop_Variability_Analysis") \
        .config("spark.sql.adaptive.enabled", "true") \
        .config("spark.sql.adaptive.coalescePartitions.enabled", "true") \
        .config("spark.sql.adaptive.skewJoin.enabled", "true") \
        .config("spark.sql.execution.arrow.pyspark.enabled", "true") \
        .config("spark.serializer", "org.apache.spark.serializer.KryoSerializer") \
        .config("spark.sql.parquet.enableVectorizedReader", "true") \
        .config("spark.sql.parquet.columnarReaderBatchSize", "8192") \
        .config("spark.sql.shuffle.partitions", "200") \
        .config("spark.sql.adaptive.advisoryPartitionSizeInBytes", "128MB") \
        .getOrCreate()
    
    # Optimiere für bessere Performance
    spark.conf.set("spark.sql.execution.arrow.maxRecordsPerBatch", "10000")
    
    print("✅ Spark Session optimiert erstellt")
    print(f"📊 Spark Version: {spark.version}")
    print(f"🔧 Konfigurierte Partitions: {spark.conf.get('spark.sql.shuffle.partitions')}")
    
    return spark

# ================================================================
# 2. SERVICE-KLASSIFIKATION UND METADATEN
# ================================================================

# Konsistente Service-Klassifikation
SERVICE_MAPPING = {
    # IPv4 Services
    '1.1.1.1': {'name': 'Cloudflare DNS', 'type': 'anycast', 'provider': 'Cloudflare', 'service_class': 'DNS'},
    '8.8.8.8': {'name': 'Google DNS', 'type': 'anycast', 'provider': 'Google', 'service_class': 'DNS'},
    '9.9.9.9': {'name': 'Quad9 DNS', 'type': 'anycast', 'provider': 'Quad9', 'service_class': 'DNS'},
    '104.16.123.96': {'name': 'Cloudflare CDN', 'type': 'anycast', 'provider': 'Cloudflare', 'service_class': 'CDN'},
    '2.16.241.219': {'name': 'Akamai CDN', 'type': 'pseudo-anycast', 'provider': 'Akamai', 'service_class': 'CDN'},
    '193.99.144.85': {'name': 'Heise', 'type': 'unicast', 'provider': 'Heise', 'service_class': 'Web'},
    '169.229.128.134': {'name': 'Berkeley NTP', 'type': 'unicast', 'provider': 'UC Berkeley', 'service_class': 'NTP'},
    
    # IPv6 Services
    '2606:4700:4700::1111': {'name': 'Cloudflare DNS', 'type': 'anycast', 'provider': 'Cloudflare', 'service_class': 'DNS'},
    '2001:4860:4860::8888': {'name': 'Google DNS', 'type': 'anycast', 'provider': 'Google', 'service_class': 'DNS'},
    '2620:fe::fe:9': {'name': 'Quad9 DNS', 'type': 'anycast', 'provider': 'Quad9', 'service_class': 'DNS'},
    '2606:4700::6810:7b60': {'name': 'Cloudflare CDN', 'type': 'anycast', 'provider': 'Cloudflare', 'service_class': 'CDN'},
    '2a02:26f0:3500:1b::1724:a393': {'name': 'Akamai CDN', 'type': 'pseudo-anycast', 'provider': 'Akamai', 'service_class': 'CDN'},
    '2a02:2e0:3fe:1001:7777:772e:2:85': {'name': 'Heise', 'type': 'unicast', 'provider': 'Heise', 'service_class': 'Web'},
    '2607:f140:ffff:8000:0:8006:0:a': {'name': 'Berkeley NTP', 'type': 'unicast', 'provider': 'UC Berkeley', 'service_class': 'NTP'}
}

# AWS-Regionen
AWS_REGIONS = {
    'eu-central-1': {'name': 'Frankfurt', 'continent': 'Europe', 'lat': 50.1109, 'lon': 8.6821},
    'eu-north-1': {'name': 'Stockholm', 'continent': 'Europe', 'lat': 59.3293, 'lon': 18.0686},
    'us-west-1': {'name': 'N. California', 'continent': 'North America', 'lat': 37.7749, 'lon': -122.4194},
    'ca-central-1': {'name': 'Canada Central', 'continent': 'North America', 'lat': 45.4215, 'lon': -75.6972},
    'af-south-1': {'name': 'Cape Town', 'continent': 'Africa', 'lat': -33.9249, 'lon': 18.4241},
    'ap-east-1': {'name': 'Hong Kong', 'continent': 'Asia', 'lat': 22.3193, 'lon': 114.1694},
    'ap-south-1': {'name': 'Mumbai', 'continent': 'Asia', 'lat': 19.0760, 'lon': 72.8777},
    'ap-northeast-1': {'name': 'Tokyo', 'continent': 'Asia', 'lat': 35.6762, 'lon': 139.6503},
    'ap-southeast-2': {'name': 'Sydney', 'continent': 'Oceania', 'lat': -33.8688, 'lon': 151.2093},
    'sa-east-1': {'name': 'São Paulo', 'continent': 'South America', 'lat': -23.5505, 'lon': -46.6333}
}

# ================================================================
# 3. PYSPARK UDFs FÜR HOP-PFAD-EXTRAKTION
# ================================================================

def extract_hop_path_spark(hubs_array):
    """
    PySpark UDF: Extrahiert Hop-Pfad aus hubs Array
    
    Args:
        hubs_array: Array von Hub-Structs (Spark Rows)
        
    Returns:
        Struct mit Pfad-Informationen
    """
    if not hubs_array or len(hubs_array) == 0:
        return {
            'ip_path': [],
            'asn_path': [],
            'hop_count': 0,
            'path_fingerprint': '',
            'valid_hops': 0
        }
    
    ip_path = []
    asn_path = []
    valid_hops = 0
    
    for hop in hubs_array:
        if not hop:
            continue
            
        try:
            # Für Spark Row-Objekte: Verwende Attribut-Zugriff statt .get()
            host = getattr(hop, 'host', None) if hasattr(hop, 'host') else None
            asn = getattr(hop, 'ASN', None) if hasattr(hop, 'ASN') else None
            loss = getattr(hop, 'Loss%', 100) if hasattr(hop, 'Loss%') else 100
            
            # Alternative: Für dict-ähnliche Zugriffe
            if host is None and hasattr(hop, '__getitem__'):
                try:
                    host = hop['host'] if 'host' in hop else None
                    asn = hop['ASN'] if 'ASN' in hop else None
                    loss = hop['Loss%'] if 'Loss%' in hop else 100
                except (KeyError, TypeError):
                    continue
            
            # Valider Hop: Host bekannt und < 100% Loss
            if host and host != '???' and host != '*' and str(host).strip() != '':
                if loss is None or loss < 100:
                    ip_path.append(str(host))
                    valid_hops += 1
                    
                    if asn and asn != '???' and str(asn).strip() != '':
                        asn_path.append(str(asn))
                        
        except Exception as e:
            # Skip problematische Hops
            continue
    
    # Pfad-Fingerprint
    path_fingerprint = '_'.join(ip_path) if ip_path else ''
    
    return {
        'ip_path': ip_path,
        'asn_path': asn_path,
        'hop_count': len(ip_path),
        'path_fingerprint': path_fingerprint,
        'valid_hops': valid_hops
    }

def calculate_hop_latency_metrics(hubs_array):
    """
    PySpark UDF: Berechnet Latenz-Metriken für Hop-Pfad
    
    Returns:
        Struct mit Latenz-Informationen
    """
    if not hubs_array or len(hubs_array) == 0:
        return {
            'end_to_end_latency': None,
            'avg_hop_latency': None,
            'max_hop_latency': None,
            'latency_variance': None,
            'packet_loss': None
        }
    
    latencies = []
    final_latency = None
    final_loss = None
    
    # Finde finalen erreichbaren Hop
    for hop in reversed(hubs_array):
        if not hop:
            continue
            
        try:
            # Für Spark Row-Objekte: Verwende Attribut-Zugriff
            avg_lat = getattr(hop, 'Avg', None) if hasattr(hop, 'Avg') else None
            loss = getattr(hop, 'Loss%', 100) if hasattr(hop, 'Loss%') else 100
            
            # Alternative: Für dict-ähnliche Zugriffe
            if avg_lat is None and hasattr(hop, '__getitem__'):
                try:
                    avg_lat = hop['Avg'] if 'Avg' in hop else None
                    loss = hop['Loss%'] if 'Loss%' in hop else 100
                except (KeyError, TypeError):
                    continue
            
            if avg_lat is not None and (loss is None or loss < 100):
                final_latency = float(avg_lat)
                final_loss = float(loss) if loss is not None else 0.0
                break
                
        except (ValueError, TypeError):
            continue
    
    # Sammle alle Hop-Latenzen
    for hop in hubs_array:
        if not hop:
            continue
            
        try:
            avg_lat = getattr(hop, 'Avg', None) if hasattr(hop, 'Avg') else None
            loss = getattr(hop, 'Loss%', 100) if hasattr(hop, 'Loss%') else 100
            
            # Alternative: Für dict-ähnliche Zugriffe
            if avg_lat is None and hasattr(hop, '__getitem__'):
                try:
                    avg_lat = hop['Avg'] if 'Avg' in hop else None
                    loss = hop['Loss%'] if 'Loss%' in hop else 100
                except (KeyError, TypeError):
                    continue
                    
            if avg_lat is not None and (loss is None or loss < 100):
                latencies.append(float(avg_lat))
                
        except (ValueError, TypeError):
            continue
    
    # Berechne Metriken
    if latencies:
        avg_latency = sum(latencies) / len(latencies)
        max_latency = max(latencies)
        if len(latencies) > 1:
            mean_lat = sum(latencies) / len(latencies)
            latency_variance = sum((x - mean_lat) ** 2 for x in latencies) / len(latencies)
        else:
            latency_variance = 0.0
    else:
        avg_latency = None
        max_latency = None
        latency_variance = None
    
    return {
        'end_to_end_latency': final_latency,
        'avg_hop_latency': avg_latency,
        'max_hop_latency': max_latency,
        'latency_variance': latency_variance,
        'packet_loss': final_loss
    }

# Definiere PySpark UDF Schemas
hop_path_schema = StructType([
    StructField("ip_path", ArrayType(StringType()), True),
    StructField("asn_path", ArrayType(StringType()), True),
    StructField("hop_count", IntegerType(), True),
    StructField("path_fingerprint", StringType(), True),
    StructField("valid_hops", IntegerType(), True)
])

latency_metrics_schema = StructType([
    StructField("end_to_end_latency", DoubleType(), True),
    StructField("avg_hop_latency", DoubleType(), True),
    StructField("max_hop_latency", DoubleType(), True),
    StructField("latency_variance", DoubleType(), True),
    StructField("packet_loss", DoubleType(), True)
])

# ================================================================
# 4. HAUPTANALYSE-FUNKTIONEN
# ================================================================

def debug_spark_data_structure(spark, df_sample):
    """
    Debug-Funktion: Inspiziert die tatsächliche Struktur der hubs-Daten
    """
    print("\n🔍 DEBUG: SPARK DATENSTRUKTUR-INSPEKTION")
    print("="*60)
    
    # Schema anzeigen
    print("📋 DATAFRAME SCHEMA:")
    df_sample.printSchema()
    
    # Sample-Daten anzeigen
    print("\n📊 SAMPLE-DATEN (erste 3 Zeilen):")
    df_sample.select("region", "dst", "hubs").show(3, truncate=False)
    
    # Hubs-Array-Struktur inspizieren
    print("\n🔬 HUBS-ARRAY-STRUKTUR:")
    try:
        # Explode hubs Array und zeige Struktur
        exploded = df_sample.select(
            "region", "dst",
            explode("hubs").alias("hop")
        ).select(
            "region", "dst", "hop.*"
        )
        
        print("Hop-Felder:")
        exploded.printSchema()
        
        print("\nSample Hop-Daten:")
        exploded.show(5)
        
    except Exception as e:
        print(f"⚠️ Fehler bei Hubs-Inspektion: {e}")
    
    # Teste UDF mit einfachem Beispiel
    print("\n🧪 TESTE VEREINFACHTE UDF:")
    try:
        def test_udf(hubs_array):
            """Einfache Test-UDF"""
            if not hubs_array:
                return "empty"
            
            try:
                hop_count = len(hubs_array)
                first_hop = hubs_array[0] if hop_count > 0 else None
                
                if first_hop:
                    # Teste verschiedene Zugriffsmethoden
                    if hasattr(first_hop, 'host'):
                        return f"attr_access_{hop_count}_hops"
                    elif hasattr(first_hop, '__getitem__') and 'host' in first_hop:
                        return f"dict_access_{hop_count}_hops"
                    else:
                        return f"unknown_access_{hop_count}_hops"
                else:
                    return f"no_hops_{hop_count}"
                    
            except Exception as e:
                return f"error_{str(e)[:20]}"
        
        test_udf_func = udf(test_udf, StringType())
        
        test_result = df_sample.withColumn(
            "test_result", test_udf_func(col("hubs"))
        ).select("region", "dst", "test_result").collect()
        
        print("UDF-Test-Ergebnisse:")
        for row in test_result[:5]:
            print(f"  {row['region']} → {row['dst']}: {row['test_result']}")
            
    except Exception as e:
        print(f"⚠️ UDF-Test fehlgeschlagen: {e}")

def load_and_enhance_data_spark(spark, file_path, protocol_name):
    """
    Lädt und erweitert MTR-Daten mit PySpark (mit Debug-Support)
    
    Args:
        spark: SparkSession
        file_path: Pfad zur Parquet-Datei
        protocol_name: 'IPv4' oder 'IPv6'
        
    Returns:
        Spark DataFrame mit erweiterten Metadaten
    """
    print(f"\n🔄 LADE UND ERWEITERE {protocol_name} DATEN MIT PYSPARK...")
    
    # Lade Parquet-Datei
    df = spark.read.parquet(file_path)
    print(f"📊 Geladene Messungen: {df.count():,}")
    
    # Sample für Debug-Zwecke
    df_sample = df.limit(10)
    
    # Debug: Inspiziere Datenstruktur
    debug_spark_data_structure(spark, df_sample)
    
    # Erstelle Service-Mapping als Broadcast Map
    service_mapping_broadcast = spark.sparkContext.broadcast(SERVICE_MAPPING)
    region_mapping_broadcast = spark.sparkContext.broadcast(AWS_REGIONS)
    
    # UDF für Service-Metadaten
    def get_service_metadata(dst):
        mapping = service_mapping_broadcast.value
        return mapping.get(dst, {
            'name': 'Unknown',
            'type': 'unknown',
            'provider': 'Unknown',
            'service_class': 'Unknown'
        })
    
    def get_region_metadata(region):
        mapping = region_mapping_broadcast.value
        return mapping.get(region, {
            'name': region,
            'continent': 'Unknown',
            'lat': 0.0,
            'lon': 0.0
        })
    
    # Registriere UDFs
    service_metadata_udf = udf(get_service_metadata, StructType([
        StructField("name", StringType(), True),
        StructField("type", StringType(), True),
        StructField("provider", StringType(), True),
        StructField("service_class", StringType(), True)
    ]))
    
    region_metadata_udf = udf(get_region_metadata, StructType([
        StructField("name", StringType(), True),
        StructField("continent", StringType(), True),
        StructField("lat", DoubleType(), True),
        StructField("lon", DoubleType(), True)
    ]))
    
    # Erweitere DataFrame
    df_enhanced = df \
        .withColumn("service_metadata", service_metadata_udf(col("dst"))) \
        .withColumn("region_metadata", region_metadata_udf(col("region"))) \
        .withColumn("service_name", col("service_metadata.name")) \
        .withColumn("service_type", col("service_metadata.type")) \
        .withColumn("provider", col("service_metadata.provider")) \
        .withColumn("service_class", col("service_metadata.service_class")) \
        .withColumn("continent", col("region_metadata.continent")) \
        .withColumn("region_name", col("region_metadata.name")) \
        .withColumn("hour", hour(col("utctime"))) \
        .withColumn("day_of_week", dayofweek(col("utctime"))) \
        .withColumn("date", F.to_date(col("utctime"))) \
        .drop("service_metadata", "region_metadata")
    
    # Bereinige Daten
    df_clean = df_enhanced.filter(
        col("hubs").isNotNull() &
        (size(col("hubs")) > 0) &
        col("region").isNotNull() &
        col("dst").isNotNull()
    )
    
    clean_count = df_clean.count()
    print(f"📊 Bereinigte Messungen: {clean_count:,} ({clean_count/df.count()*100:.1f}%)")
    
    return df_clean

def extract_paths_parallel_spark_alternative(spark, df_clean, protocol_name):
    """
    ALTERNATIVE: Spark SQL-basierte Pfad-Extraktion (falls UDFs Probleme haben)
    
    Args:
        spark: SparkSession
        df_clean: Bereinigter Spark DataFrame
        protocol_name: 'IPv4' oder 'IPv6'
        
    Returns:
        Spark DataFrame mit extrahierten Pfad-Informationen
    """
    print(f"\n🛤️ ALTERNATIVE SPARK SQL PFAD-EXTRAKTION FÜR {protocol_name}...")
    
    # Erstelle temporäre View
    df_clean.createOrReplaceTempView("mtr_data")
    
    # Spark SQL für Pfad-Extraktion
    paths_df = spark.sql("""
        SELECT 
            id as measurement_id,
            region,
            dst,
            service_name,
            service_type,
            provider,
            service_class,
            continent,
            utctime as timestamp,
            date,
            size(hubs) as hop_count,
            -- Vereinfachte Pfad-Extraktion
            concat_ws('_', 
                transform(
                    filter(hubs, h -> h.host IS NOT NULL AND h.host != '???' AND h.`Loss%` < 100),
                    h -> h.host
                )
            ) as path_fingerprint,
            -- End-to-End Latenz (letzter valider Hop)
            (
                SELECT h.Avg 
                FROM (
                    SELECT explode(reverse(hubs)) as h
                ) 
                WHERE h.host IS NOT NULL AND h.host != '???' AND h.`Loss%` < 100 AND h.Avg IS NOT NULL
                LIMIT 1
            ) as end_to_end_latency,
            -- Packet Loss (letzter valider Hop)
            (
                SELECT h.`Loss%`
                FROM (
                    SELECT explode(reverse(hubs)) as h
                ) 
                WHERE h.host IS NOT NULL AND h.host != '???' AND h.`Loss%` < 100
                LIMIT 1
            ) as packet_loss,
            -- Anzahl valider Hops
            size(
                filter(hubs, h -> h.host IS NOT NULL AND h.host != '???' AND h.`Loss%` < 100)
            ) as valid_hops
        FROM mtr_data
        WHERE size(hubs) > 0
    """)
    
    # Filtere nur valide Pfade
    paths_df_filtered = paths_df.filter(
        (col("hop_count") > 0) & 
        (col("valid_hops") > 0) &
        (col("path_fingerprint") != "")
    )
    
    # Cache für Performance
    paths_df_filtered.cache()
    
    path_count = paths_df_filtered.count()
    print(f"✅ Extrahierte Pfade (SQL): {path_count:,}")
    
    return paths_df_filtered

def extract_paths_parallel_spark(spark, df_clean, protocol_name):
    """
    Parallele Hop-Pfad-Extraktion mit PySpark (mit Fallback)
    
    Args:
        spark: SparkSession
        df_clean: Bereinigter Spark DataFrame
        protocol_name: 'IPv4' oder 'IPv6'
        
    Returns:
        Spark DataFrame mit extrahierten Pfad-Informationen
    """
    print(f"\n🛤️ PARALLELE HOP-PFAD-EXTRAKTION FÜR {protocol_name}...")
    
    try:
        # Versuche UDF-basierte Extraktion
        print("🔧 Versuche UDF-basierte Pfad-Extraktion...")
        
        # Registriere UDFs
        extract_path_udf = udf(extract_hop_path_spark, hop_path_schema)
        extract_latency_udf = udf(calculate_hop_latency_metrics, latency_metrics_schema)
        
        # Extrahiere Pfade parallel
        paths_df = df_clean \
            .withColumn("path_info", extract_path_udf(col("hubs"))) \
            .withColumn("latency_info", extract_latency_udf(col("hubs"))) \
            .select(
                col("id").alias("measurement_id"),
                col("region"),
                col("dst"),
                col("service_name"),
                col("service_type"),
                col("provider"),
                col("service_class"),
                col("continent"),
                col("utctime").alias("timestamp"),
                col("date"),
                col("path_info.ip_path"),
                col("path_info.asn_path"),
                col("path_info.hop_count"),
                col("path_info.path_fingerprint"),
                col("path_info.valid_hops"),
                col("latency_info.end_to_end_latency"),
                col("latency_info.avg_hop_latency"),
                col("latency_info.max_hop_latency"),
                col("latency_info.latency_variance"),
                col("latency_info.packet_loss")
            ) \
            .filter(col("hop_count") > 0)  # Nur valide Pfade
        
        # Test ob UDFs funktionieren
        paths_df.cache()
        path_count = paths_df.count()
        
        print(f"✅ UDF-Extraktion erfolgreich: {path_count:,} Pfade")
        return paths_df
        
    except Exception as e:
        print(f"⚠️ UDF-Extraktion fehlgeschlagen: {e}")
        print("🔄 Fallback zu Spark SQL-Extraktion...")
        
        # Fallback zu SQL-basierter Extraktion
        return extract_paths_parallel_spark_alternative(spark, df_clean, protocol_name)

def calculate_variability_metrics_spark_simplified(spark, paths_df, protocol_name):
    """
    Vereinfachte Variabilität-Metriken mit Spark SQL (robuster als UDFs)
    
    Args:
        spark: SparkSession
        paths_df: DataFrame mit Pfad-Informationen
        protocol_name: 'IPv4' oder 'IPv6'
        
    Returns:
        Spark DataFrame mit Variabilität-Metriken
    """
    print(f"\n📈 BERECHNE VARIABILITÄT-METRIKEN (VEREINFACHT) FÜR {protocol_name}...")
    
    # Debug: Zeige verfügbare Spalten
    print(f"📋 Verfügbare Spalten: {paths_df.columns}")
    
    # Erstelle temporäre View
    paths_df.createOrReplaceTempView("paths_data")
    
    # Grundlegende Aggregationen mit Spark SQL - ALLE SPALTEN BEIBEHALTEN
    variability_df = spark.sql("""
        SELECT 
            region,
            dst,
            ANY_VALUE(service_name) as service_name,
            ANY_VALUE(service_type) as service_type,
            ANY_VALUE(provider) as provider,
            ANY_VALUE(service_class) as service_class,  -- KRITISCH: service_class hinzufügen
            ANY_VALUE(continent) as continent,
            COUNT(*) as measurement_count,
            COUNT(DISTINCT path_fingerprint) as unique_paths,
            COUNT(DISTINCT date) as analyzed_days,
            AVG(hop_count) as avg_hop_count,
            STDDEV(hop_count) as hop_count_stddev,
            AVG(end_to_end_latency) as avg_latency,
            MIN(timestamp) as first_measurement,
            MAX(timestamp) as last_measurement,
            -- Stabilität-Metriken (vereinfacht)
            ROUND(
                (1.0 - CAST(COUNT(DISTINCT path_fingerprint) AS DOUBLE) / COUNT(*)) * 100, 2
            ) as ip_stability,
            ROUND(
                (CAST(COUNT(DISTINCT path_fingerprint) AS DOUBLE) / COUNT(*)) * 100, 2
            ) as path_diversity,
            -- Dominante-Pfad-Ratio (vereinfacht - bester Pfad)
            ROUND(
                (
                    SELECT CAST(COUNT(*) AS DOUBLE) 
                    FROM (
                        SELECT path_fingerprint, COUNT(*) as cnt
                        FROM paths_data p2 
                        WHERE p2.region = p1.region AND p2.dst = p1.dst
                        GROUP BY path_fingerprint 
                        ORDER BY cnt DESC 
                        LIMIT 1
                    )
                ) / COUNT(*) * 100, 2
            ) as dominant_path_ratio,
            -- Hop-Count-Variabilität
            ROUND(
                CASE 
                    WHEN AVG(hop_count) > 0 
                    THEN (STDDEV(hop_count) / AVG(hop_count)) * 100 
                    ELSE 0 
                END, 2
            ) as hop_count_cv
        FROM paths_data p1
        GROUP BY region, dst
        HAVING COUNT(*) >= 5  -- Mindestens 5 Messungen
    """)
    
    # Erweitere um zeitliche Metriken
    final_df = variability_df \
        .withColumn("timespan_days", 
            F.datediff(col("last_measurement"), col("first_measurement"))) \
        .withColumn("temporal_stability", 
            F.when(col("hop_count_cv") > 0, 100 - col("hop_count_cv")).otherwise(100)) \
        .withColumn("route_consistency", col("dominant_path_ratio"))
    
    # Debug: Zeige finale Spalten
    print(f"📋 Finale Spalten: {final_df.columns}")
    
    # Cache Ergebnisse
    final_df.cache()
    
    pair_count = final_df.count()
    print(f"✅ Analysierte Messpunkt-Service-Paare (vereinfacht): {pair_count}")
    
    return final_df

def calculate_variability_metrics_spark(spark, paths_df, protocol_name):
    """
    Berechnet Variabilität-Metriken pro Messpunkt-Zielservice-Paar mit PySpark (mit Fallback)
    
    Args:
        spark: SparkSession
        paths_df: DataFrame mit Pfad-Informationen
        protocol_name: 'IPv4' oder 'IPv6'
        
    Returns:
        Spark DataFrame mit Variabilität-Metriken
    """
    print(f"\n📈 BERECHNE VARIABILITÄT-METRIKEN PRO PAAR ({protocol_name})...")
    
    try:
        # Versuche komplexe UDF-basierte Berechnung
        print("🔧 Versuche erweiterte Variabilität-UDFs...")
        
        # Gruppe nach Region-Destination-Paaren
        pair_groups = paths_df.groupBy("region", "dst", "service_name", "service_type", "provider", "continent")
        
        # Grundlegende Aggregationen
        basic_agg = pair_groups.agg(
            count("*").alias("measurement_count"),
            F.countDistinct("path_fingerprint").alias("unique_paths"),
            F.countDistinct("date").alias("analyzed_days"),
            avg("hop_count").alias("avg_hop_count"),
            stddev("hop_count").alias("hop_count_stddev"),
            avg("end_to_end_latency").alias("avg_latency"),
            F.min("timestamp").alias("first_measurement"),
            F.max("timestamp").alias("last_measurement"),
            collect_list("path_fingerprint").alias("all_paths"),
            collect_list("hop_count").alias("all_hop_counts")
        ).filter(col("measurement_count") >= 5)  # Mindestens 5 Messungen
        
        # UDF für erweiterte Stabilität-Metriken
        def calculate_path_stability_advanced(all_paths, measurement_count, unique_paths):
            """Berechnet erweiterte Stabilität-Metriken"""
            if not all_paths or measurement_count < 2:
                return {
                    'ip_stability': 100.0,
                    'path_diversity': 0.0,
                    'dominant_path_ratio': 100.0,
                    'route_consistency': 100.0
                }
            
            # Pfad-Diversität
            path_diversity = (unique_paths / measurement_count) * 100
            
            # Dominante-Pfad-Ratio
            from collections import Counter
            path_counts = Counter(all_paths)
            most_common_count = max(path_counts.values()) if path_counts else measurement_count
            dominant_path_ratio = (most_common_count / measurement_count) * 100
            
            # IP-Stabilität (inverse Diversität)
            ip_stability = 100 - path_diversity
            
            # Route-Konsistenz (Anteil der häufigsten Route)
            route_consistency = dominant_path_ratio
            
            return {
                'ip_stability': float(ip_stability),
                'path_diversity': float(path_diversity),
                'dominant_path_ratio': float(dominant_path_ratio),
                'route_consistency': float(route_consistency)
            }
        
        def calculate_temporal_stability(all_hop_counts):
            """Berechnet zeitliche Hop-Count-Stabilität"""
            if not all_hop_counts or len(all_hop_counts) < 2:
                return {
                    'hop_count_cv': 0.0,
                    'temporal_stability': 100.0
                }
            
            # Hop-Count-Variationskoeffizient
            try:
                import statistics
                mean_hops = statistics.mean(all_hop_counts)
                std_hops = statistics.stdev(all_hop_counts) if len(all_hop_counts) > 1 else 0
                hop_count_cv = (std_hops / mean_hops) * 100 if mean_hops > 0 else 0.0
            except:
                hop_count_cv = 0.0
            
            # Temporale Stabilität (inverse CV)
            temporal_stability = max(0, 100 - hop_count_cv)
            
            return {
                'hop_count_cv': float(hop_count_cv),
                'temporal_stability': float(temporal_stability)
            }
        
        # Registriere erweiterte UDFs
        stability_schema = StructType([
            StructField("ip_stability", DoubleType(), True),
            StructField("path_diversity", DoubleType(), True),
            StructField("dominant_path_ratio", DoubleType(), True),
            StructField("route_consistency", DoubleType(), True)
        ])
        
        temporal_schema = StructType([
            StructField("hop_count_cv", DoubleType(), True),
            StructField("temporal_stability", DoubleType(), True)
        ])
        
        path_stability_udf = udf(calculate_path_stability_advanced, stability_schema)
        temporal_stability_udf = udf(calculate_temporal_stability, temporal_schema)
        
        # Erweiterte Metriken berechnen
        variability_df = basic_agg \
            .withColumn("stability_metrics", path_stability_udf(
                col("all_paths"), 
                col("measurement_count"), 
                col("unique_paths")
            )) \
            .withColumn("temporal_metrics", temporal_stability_udf(col("all_hop_counts"))) \
            .withColumn("timespan_days", 
                F.datediff(col("last_measurement"), col("first_measurement"))) \
            .select(
                col("region"),
                col("dst"),
                col("service_name"),
                col("service_type"),
                col("provider"),
                col("continent"),
                col("measurement_count"),
                col("unique_paths"),
                col("analyzed_days"),
                col("timespan_days"),
                col("avg_hop_count"),
                col("hop_count_stddev"),
                col("avg_latency"),
                col("stability_metrics.ip_stability"),
                col("stability_metrics.path_diversity"),
                col("stability_metrics.dominant_path_ratio"),
                col("stability_metrics.route_consistency"),
                col("temporal_metrics.hop_count_cv"),
                col("temporal_metrics.temporal_stability")
            )
        
        # Cache Ergebnisse
        variability_df.cache()
        
        # Test ob UDFs funktionieren
        pair_count = variability_df.count()
        print(f"✅ UDF-basierte Variabilität erfolgreich: {pair_count} Paare")
        
        return variability_df
        
    except Exception as e:
        print(f"⚠️ UDF-basierte Variabilität fehlgeschlagen: {e}")
        print("🔄 Fallback zu vereinfachter SQL-Berechnung...")
        
        # Fallback zu vereinfachter SQL-basierter Berechnung
        return calculate_variability_metrics_spark_simplified(spark, paths_df, protocol_name)

def calculate_summary_statistics_spark(variability_df):
    """Berechnet zusammenfassende Statistiken mit Spark"""
    
    # Grundlegende Statistiken
    total_pairs = variability_df.count()
    
    summary_stats = variability_df.agg(
        avg("ip_stability").alias("avg_ip_stability"),
        avg("path_diversity").alias("avg_path_diversity"),
        avg("hop_count_cv").alias("avg_hop_count_cv"),
        F.max("ip_stability").alias("max_stability"),
        F.min("ip_stability").alias("min_stability")
    ).collect()[0]
    
    # Service-Type-Statistiken
    service_type_stats = variability_df.groupBy("service_type").agg(
        count("*").alias("count"),
        avg("ip_stability").alias("avg_stability")
    ).collect()
    
    # Provider-Statistiken
    provider_stats = variability_df.groupBy("provider").agg(
        count("*").alias("count"),
        avg("ip_stability").alias("avg_stability")
    ).orderBy(F.desc("avg_stability")).collect()
    
    # Kategorisierung
    high_variability = variability_df.filter(col("ip_stability") < 50).count()
    stable_pairs = variability_df.filter(col("ip_stability") > 80).count()
    
    # Top/Bottom Paare
    top_stable = variability_df.orderBy(F.desc("ip_stability")).limit(5).collect()
    bottom_stable = variability_df.orderBy("ip_stability").limit(5).collect()
    
    return {
        'total_pairs': total_pairs,
        'avg_ip_stability': summary_stats['avg_ip_stability'],
        'avg_path_diversity': summary_stats['avg_path_diversity'],
        'avg_hop_count_cv': summary_stats['avg_hop_count_cv'],
        'max_stability': summary_stats['max_stability'],
        'min_stability': summary_stats['min_stability'],
        'high_variability_pairs': high_variability,
        'stable_pairs': stable_pairs,
        'service_type_stats': {row['service_type']: row.asDict() for row in service_type_stats},
        'provider_stats': {row['provider']: row.asDict() for row in provider_stats},
        'top_stable_pairs': [row.asDict() for row in top_stable],
        'bottom_stable_pairs': [row.asDict() for row in bottom_stable]
    }

# ================================================================
# 5. STATISTISCHE VALIDIERUNG MIT SPARK
# ================================================================

def statistical_validation_spark(spark, variability_df):
    """Statistische Validierung mit Spark und Pandas für komplexe Tests"""
    
    print("\n📊 STATISTISCHE VALIDIERUNG (SPARK + PANDAS HYBRID)...")
    
    # Konvertiere zu Pandas für erweiterte statistische Tests
    pandas_df = variability_df.toPandas()
    
    validation_results = {}
    
    # Service-Type-Vergleiche
    anycast_data = pandas_df[pandas_df['service_type'] == 'anycast']['ip_stability']
    unicast_data = pandas_df[pandas_df['service_type'] == 'unicast']['ip_stability']
    
    if len(anycast_data) > 0 and len(unicast_data) > 0:
        # Bereinige Daten
        anycast_clean = anycast_data.dropna()
        unicast_clean = unicast_data.dropna()
        
        if len(anycast_clean) >= 3 and len(unicast_clean) >= 3:
            # Mann-Whitney U Test
            u_stat, p_value = stats.mannwhitneyu(anycast_clean, unicast_clean, alternative='two-sided')
            
            # Effect Size (Cliff's Delta)
            def calculate_cliffs_delta(group1, group2):
                n1, n2 = len(group1), len(group2)
                dominance = sum(x > y for x in group1 for y in group2) - sum(x < y for x in group1 for y in group2)
                return dominance / (n1 * n2)
            
            effect_size = calculate_cliffs_delta(anycast_clean, unicast_clean)
            
            validation_results['anycast_vs_unicast'] = {
                'anycast_mean': float(anycast_clean.mean()),
                'unicast_mean': float(unicast_clean.mean()),
                'p_value': float(p_value),
                'cliffs_delta': float(effect_size),
                'significant': p_value < 0.05,
                'anycast_n': len(anycast_clean),
                'unicast_n': len(unicast_clean)
            }
    
    return validation_results

# ================================================================
# 6. VISUALISIERUNGEN (PANDAS-HYBRID)
# ================================================================

def create_visualizations_hybrid(variability_df, summary_stats, validation_results, protocol_name):
    """Erstellt Visualisierungen mit Pandas-Hybrid-Ansatz (robust gegen fehlende Spalten)"""
    
    print(f"\n🎨 ERSTELLE VISUALISIERUNGEN FÜR {protocol_name}...")
    
    # Konvertiere zu Pandas für Plotting
    df = variability_df.toPandas()
    
    # Debug: Zeige verfügbare Spalten
    print(f"📋 Verfügbare Spalten für Plotting: {list(df.columns)}")
    
    fig = plt.figure(figsize=(24, 20))
    
    # 1. Stabilität Heatmap
    plt.subplot(4, 4, 1)
    if len(df) > 0 and 'service_name' in df.columns and 'region' in df.columns and 'ip_stability' in df.columns:
        try:
            pivot = df.pivot_table(values='ip_stability', index='region', columns='service_name', aggfunc='mean')
            sns.heatmap(pivot, annot=True, fmt='.1f', cmap='RdYlGn', vmin=0, vmax=100, cbar_kws={'label': 'Stabilität (%)'})
        except Exception as e:
            plt.text(0.5, 0.5, f'Heatmap Fehler:\n{str(e)}', ha='center', va='center', transform=plt.gca().transAxes)
    else:
        plt.text(0.5, 0.5, 'Daten für Heatmap\nnicht verfügbar', ha='center', va='center', transform=plt.gca().transAxes)
    plt.title(f'{protocol_name}: Stabilität Heatmap\nMesspunkt × Service', fontweight='bold')
    
    # 2. Provider Ranking
    plt.subplot(4, 4, 2)
    if 'provider_stats' in summary_stats and summary_stats['provider_stats']:
        try:
            providers = list(summary_stats['provider_stats'].keys())[:6]  # Top 6
            stabilities = [summary_stats['provider_stats'][p]['avg_stability'] for p in providers]
            
            bars = plt.barh(range(len(providers)), stabilities, color=['green' if s > 80 else 'orange' if s > 60 else 'red' for s in stabilities])
            plt.yticks(range(len(providers)), providers)
            plt.xlabel('Ø Stabilität (%)')
            plt.title(f'{protocol_name}: Provider Rankings', fontweight='bold')
            plt.xlim(0, 100)
        except Exception as e:
            plt.text(0.5, 0.5, f'Provider Stats Fehler:\n{str(e)}', ha='center', va='center', transform=plt.gca().transAxes)
    else:
        plt.text(0.5, 0.5, 'Provider-Daten\nnicht verfügbar', ha='center', va='center', transform=plt.gca().transAxes)
    plt.title(f'{protocol_name}: Provider Rankings', fontweight='bold')
    
    # 3. Service Type Verteilungen
    plt.subplot(4, 4, 3)
    if 'service_type' in df.columns and 'ip_stability' in df.columns:
        try:
            for stype in df['service_type'].unique():
                data = df[df['service_type'] == stype]['ip_stability']
                plt.hist(data, alpha=0.7, label=f'{stype} (n={len(data)})', bins=15)
            plt.xlabel('IP-Stabilität (%)')
            plt.ylabel('Anzahl Paare')
            plt.legend()
        except Exception as e:
            plt.text(0.5, 0.5, f'Verteilungs-Fehler:\n{str(e)}', ha='center', va='center', transform=plt.gca().transAxes)
    else:
        plt.text(0.5, 0.5, 'Service-Type-Daten\nnicht verfügbar', ha='center', va='center', transform=plt.gca().transAxes)
    plt.title(f'{protocol_name}: Service-Type-Verteilungen', fontweight='bold')
    
    # 4. Kontinentale Patterns
    plt.subplot(4, 4, 4)
    if 'continent' in df.columns and 'ip_stability' in df.columns:
        try:
            continent_stats = df.groupby('continent')['ip_stability'].agg(['mean', 'std'])
            plt.bar(range(len(continent_stats)), continent_stats['mean'], 
                    yerr=continent_stats['std'], capsize=5, alpha=0.8)
            plt.xticks(range(len(continent_stats)), continent_stats.index, rotation=45)
            plt.ylabel('Ø Stabilität (%)')
        except Exception as e:
            plt.text(0.5, 0.5, f'Kontinental-Fehler:\n{str(e)}', ha='center', va='center', transform=plt.gca().transAxes)
    else:
        plt.text(0.5, 0.5, 'Kontinental-Daten\nnicht verfügbar', ha='center', va='center', transform=plt.gca().transAxes)
    plt.title(f'{protocol_name}: Kontinentale Patterns', fontweight='bold')
    
    # 5-12: Weitere Plots mit robuster Fehlerbehandlung
    plot_configs = [
        ('Diversität vs. Stabilität', 'path_diversity', 'ip_stability'),
        ('Hop-CV vs. Stabilität', 'hop_count_cv', 'ip_stability'),
        ('Temporal Stabilität', 'temporal_stability', 'ip_stability'),
        ('Route Konsistenz', 'route_consistency', 'ip_stability'),
        ('Messung Count vs. Stabilität', 'measurement_count', 'ip_stability'),
        ('Timespan vs. Stabilität', 'timespan_days', 'ip_stability'),
        ('Service Class Performance', None, None),  # Bar chart
        ('Top/Bottom Rankings', None, None)  # Horizontal bar
    ]
    
    for i, (title, x_col, y_col) in enumerate(plot_configs, 5):
        plt.subplot(4, 4, i)
        
        try:
            if x_col and y_col and x_col in df.columns and y_col in df.columns:
                # Scatter plots
                service_colors = {'anycast': 'red', 'unicast': 'blue', 'pseudo-anycast': 'orange'}
                if 'service_type' in df.columns:
                    colors = df['service_type'].map(service_colors).fillna('gray')
                else:
                    colors = 'blue'
                
                plt.scatter(df[x_col], df[y_col], alpha=0.7, c=colors)
                plt.xlabel(x_col.replace('_', ' ').title())
                plt.ylabel(y_col.replace('_', ' ').title())
            else:
                # Custom plots für spezielle Fälle
                if 'Service Class' in title:
                    if 'service_class' in df.columns and 'ip_stability' in df.columns:
                        service_class_stats = df.groupby('service_class')['ip_stability'].mean().sort_values(ascending=False)
                        plt.bar(range(len(service_class_stats)), service_class_stats.values, alpha=0.8)
                        plt.xticks(range(len(service_class_stats)), service_class_stats.index, rotation=45)
                        plt.ylabel('Ø Stabilität (%)')
                    else:
                        plt.text(0.5, 0.5, 'service_class\nSpalte fehlt', ha='center', va='center', transform=plt.gca().transAxes)
                elif 'Rankings' in title and 'top_stable_pairs' in summary_stats:
                    # Top 5 und Bottom 5
                    top_5 = summary_stats['top_stable_pairs'][:5]
                    bottom_5 = summary_stats['bottom_stable_pairs'][:5]
                    
                    if top_5 and bottom_5:
                        labels = [f"{p['region']} → {p['service_name'][:15]}" for p in top_5 + bottom_5]
                        values = [p['ip_stability'] for p in top_5 + bottom_5]
                        colors = ['green'] * len(top_5) + ['red'] * len(bottom_5)
                        
                        if len(labels) <= 10:  # Sicherstellen dass wir nicht zu viele haben
                            plt.barh(range(len(labels)), values, color=colors, alpha=0.7)
                            plt.yticks(range(len(labels)), labels, fontsize=8)
                            plt.xlabel('Stabilität (%)')
                    else:
                        plt.text(0.5, 0.5, 'Ranking-Daten\nnicht verfügbar', ha='center', va='center', transform=plt.gca().transAxes)
                else:
                    plt.text(0.5, 0.5, f'Daten für {title}\nnicht verfügbar', ha='center', va='center', transform=plt.gca().transAxes)
        
        except Exception as e:
            plt.text(0.5, 0.5, f'Plot-Fehler:\n{str(e)[:30]}...', ha='center', va='center', transform=plt.gca().transAxes)
        
        plt.title(f'{protocol_name}: {title}', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig(f'pyspark_hop_variability_{protocol_name.lower()}.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    print(f"✅ {protocol_name}: 12 Visualisierungen erstellt (robust gegen fehlende Spalten)")


# ================================================================
# 7. HAUPTANALYSE-FUNKTION
# ================================================================

def run_pyspark_hop_variability_analysis():
    """Hauptfunktion: PySpark-optimierte Hop-Variabilität-Analyse"""
    
    print("🚀 STARTE PYSPARK-OPTIMIERTE HOP-VARIABILITÄT-ANALYSE...")
    
    # Dateipfade
    IPv4_FILE = "../data/IPv4.parquet"  # Anpassen
    IPv6_FILE = "../data/IPv6.parquet"  # Anpassen
    
    # Spark Session
    spark = create_optimized_spark_session()
    
    try:
        results = {}
        validation_results = {}
        
        # Analysiere beide Protokolle
        for file_path, protocol in [(IPv4_FILE, 'IPv4'), (IPv6_FILE, 'IPv6')]:
            print(f"\n{'='*80}")
            print(f"PYSPARK ANALYSE: {protocol}")
            print("="*80)
            
            # 1. Daten laden und erweitern
            df_clean = load_and_enhance_data_spark(spark, file_path, protocol)
            
            # 2. Parallele Pfad-Extraktion
            paths_df = extract_paths_parallel_spark(spark, df_clean, protocol)
            
            # 3. Variabilität-Metriken berechnen
            variability_df = calculate_variability_metrics_spark(spark, paths_df, protocol)
            
            # 4. Zusammenfassende Statistiken
            summary_stats = calculate_summary_statistics_spark(variability_df)
            
            # 5. Statistische Validierung
            validation = statistical_validation_spark(spark, variability_df)
            
            # 6. Visualisierungen
            create_visualizations_hybrid(variability_df, summary_stats, validation, protocol)
            
            # Speichere Ergebnisse
            results[protocol] = {
                'variability_df': variability_df,
                'summary_stats': summary_stats,
                'protocol': protocol
            }
            validation_results[protocol] = validation
        
        # ================================================================
        # ZUSAMMENFASSUNG
        # ================================================================
        
        print(f"\n{'='*120}")
        print("🎯 PYSPARK HOP-VARIABILITÄT-ANALYSE ABGESCHLOSSEN")
        print("="*120)
        
        for protocol in ['IPv4', 'IPv6']:
            if protocol in results:
                summary = results[protocol]['summary_stats']
                validation = validation_results.get(protocol, {})
                
                print(f"\n📊 {protocol} PERFORMANCE-ZUSAMMENFASSUNG:")
                print(f"  🔢 Analysierte Paare: {summary['total_pairs']}")
                print(f"  📈 Ø IP-Stabilität: {summary['avg_ip_stability']:.1f}%")
                print(f"  🎯 Stabilste Paare (>80%): {summary['stable_pairs']}")
                print(f"  ⚠️ Variable Paare (<50%): {summary['high_variability_pairs']}")
                
                if 'anycast_vs_unicast' in validation:
                    val = validation['anycast_vs_unicast']
                    print(f"\n🔬 STATISTISCHE VALIDIERUNG:")
                    print(f"  Anycast vs. Unicast Stabilität: {val['anycast_mean']:.1f}% vs. {val['unicast_mean']:.1f}%")
                    print(f"  Signifikant: {'✅' if val['significant'] else '❌'} (p={val['p_value']:.4f})")
                    print(f"  Effect Size: {val['cliffs_delta']:.3f}")
        
        print(f"\n🏆 PYSPARK-OPTIMIERUNGEN ERFOLGREICH:")
        optimizations = [
            "✅ Parallele Pfad-Extraktion mit UDFs",
            "✅ Spark-native Aggregationen und Gruppierungen", 
            "✅ Broadcast-optimierte Service-Mappings",
            "✅ Adaptive Query Execution (AQE) aktiviert",
            "✅ Columnar Parquet-Reading optimiert",
            "✅ Memory-effiziente DataFrame-Operationen",
            "✅ Pandas-Hybrid für komplexe Statistiken",
            "✅ 12+ wissenschaftliche Visualisierungen pro Protokoll"
        ]
        
        for opt in optimizations:
            print(f"  {opt}")
        
        return results, validation_results
        
    except Exception as e:
        print(f"❌ Fehler in PySpark-Analyse: {e}")
        import traceback
        traceback.print_exc()
        return None, None
        
    finally:
        # Spark Session beenden
        spark.stop()
        print("🔥 Spark Session beendet")

# ================================================================
# 8. PERFORMANCE-VERGLEICHSFUNKTIONEN
# ================================================================

def benchmark_pandas_vs_pyspark():
    """Benchmark: Pandas vs. PySpark Performance-Vergleich"""
    
    print("\n⏱️ PERFORMANCE-BENCHMARK: PANDAS VS. PYSPARK")
    print("="*60)
    
    import time
    
    # Erstelle Test-Dataset
    print("📊 Erstelle Benchmark-Dataset...")
    
    # Simuliere größeres Dataset für aussagekräftigen Vergleich
    test_data = {
        'region': ['eu-central-1', 'us-west-1', 'ap-east-1'] * 10000,
        'dst': ['1.1.1.1', '8.8.8.8', '9.9.9.9'] * 10000,
        'path_fingerprint': [f'path_{i%100}' for i in range(30000)],
        'hop_count': np.random.randint(5, 20, 30000),
        'ip_stability': np.random.uniform(20, 95, 30000)
    }
    
    # Pandas Benchmark
    start_pandas = time.time()
    pandas_df = pd.DataFrame(test_data)
    pandas_grouped = pandas_df.groupby(['region', 'dst']).agg({
        'ip_stability': ['mean', 'std', 'count'],
        'hop_count': 'mean',
        'path_fingerprint': 'nunique'
    })
    pandas_time = time.time() - start_pandas
    
    # PySpark Benchmark
    spark = create_optimized_spark_session()
    start_spark = time.time()
    spark_df = spark.createDataFrame(pd.DataFrame(test_data))
    spark_grouped = spark_df.groupBy('region', 'dst').agg(
        F.avg('ip_stability').alias('avg_stability'),
        F.stddev('ip_stability').alias('std_stability'),
        F.count('*').alias('count'),
        F.avg('hop_count').alias('avg_hops'),
        F.countDistinct('path_fingerprint').alias('unique_paths')
    ).collect()
    spark_time = time.time() - start_spark
    
    spark.stop()
    
    print(f"📈 BENCHMARK-ERGEBNISSE:")
    print(f"  Pandas Zeit: {pandas_time:.3f}s")
    print(f"  PySpark Zeit: {spark_time:.3f}s")
    print(f"  Speedup: {pandas_time/spark_time:.2f}x {'(PySpark schneller)' if spark_time < pandas_time else '(Pandas schneller)'}")
    print(f"  Dataset-Größe: {len(test_data['region']):,} Zeilen")

# ================================================================
# 9. AUSFÜHRUNG
# ================================================================

if __name__ == "__main__":
    print("="*120)
    print("📋 ANWEISUNGEN FÜR PYSPARK HOP-VARIABILITÄT-ANALYSE:")
    print("="*120)
    print("1. Passen Sie die Dateipfade IPv4_FILE und IPv6_FILE an")
    print("2. Stellen Sie sicher, dass PySpark installiert ist: pip install pyspark")
    print("3. Führen Sie run_pyspark_hop_variability_analysis() aus")
    print("4. Für Performance-Vergleich: benchmark_pandas_vs_pyspark()")
    print("5. OPTIMIERT: Parallele UDFs, Spark SQL, Broadcast-Maps, AQE")
    print("6. HYBRID: Spark für Big Data + Pandas für erweiterte Statistiken")
    print("="*120)
    
    # Benchmark optional ausführen
    print("\n🔥 Führe Performance-Benchmark aus...")
    benchmark_pandas_vs_pyspark()
    
    # Hauptanalyse
    print("\n🚀 Führe PySpark-optimierte Analyse aus...")
    results, validation = run_pyspark_hop_variability_analysis()
```
